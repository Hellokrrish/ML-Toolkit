```python
# (iii) Implement a function to calculate the weighted mean of a list of values and their corresponding weights.
import statistics

def weighted_mean(values, weights):
    """
    Calculate the weighted mean of a list of values and their corresponding weights.

    :param values: List of values
    :param weights: List of corresponding weights
    :return: Weighted mean
    """
    if len(values) != len(weights):
        raise ValueError("The lengths of values and weights must be the same")

    weighted_sum = statistics.fsum(value * weight for value, weight in zip(values, weights))
    total_weight = statistics.fsum(weights)
    if total_weight == 0:
        raise ValueError("The sum of weights must not be zero")
    return weighted_sum / total_weight
values = [10, 20, 30, 40]
weights = [1, 2, 3, 4]
print(f"The weighted mean is: {weighted_mean(values, weights)}")
```

```python
#  (iv) Write a Python function to find the geometric mean of a list of positive numbers.
import math

def geometric_mean(numbers):
    # Ensure the list is not empty
    if len(numbers) == 0:
        return 0
    
    product = math.prod(numbers)
    n = len(numbers)
    geom_mean = product ** (1/n)
    return geom_mean
numbers = [1, 2, 3, 4, 5]
print(f"The geometric mean of {numbers} is {geometric_mean(numbers)}")
```

```python
#  (v) Create a program to calculate the harmonic mean of a list of values.
def harmonic_mean(numbers):
    # Ensure the list is not empty
    if len(numbers) == 0:
        return 0
    
    # Calculate the sum of reciprocals of the numbers
    reciprocal_sum = sum(1 / x for x in numbers if x != 0)
    
    # Calculate the harmonic mean
    n = len(numbers)
    if reciprocal_sum == 0:
        return 0
    
    harm_mean = n / reciprocal_sum
    
    return harm_mean

# Example usage:
numbers = [1, 2, 3, 4, 5]
print(f"The harmonic mean of {numbers} is {harmonic_mean(numbers)}")
```

```python
#  (vii) Implement a Python program to find the trimmed mean of a list, excluding a certain percentage of outliers.
def trimmed_mean(numbers, percentage):
    if len(numbers) == 0:
        raise ValueError("The list is empty, cannot compute trimmed mean")   
    if percentage < 0 or percentage >= 50:
        raise ValueError("Percentage must be between 0 and 50")    
    # Sort the list
    sorted_numbers = sorted(numbers)  
    # Calculate the number of elements to trim from each end
    n = len(numbers)
    trim_count = int(n * percentage / 100)   
    # Trim the numbers
    trimmed_numbers = sorted_numbers[trim_count:n - trim_count]    
    # Calculate the mean of the trimmed list
    if len(trimmed_numbers) == 0:        raise ValueError("Trimmed list is empty, cannot compute trimmed mean")    
    trimmed_mean_value = sum(trimmed_numbers) / len(trimmed_numbers)   
    return trimmed_mean_value
# Example usage:
numbers = [1, 2, 3, 4, 5, 100]
percentage = 20
print(f"The trimmed mean of {numbers} with {percentage}% trimmed is {trimmed_mean(numbers, percentage)}")
```

### 3)  Write a Python class representing a discrete random variable with methods to calculate its expected value and variance.

```python
class DiscreteRandomVariable:
    def __init__(self, values, probabilities):
        """
        Initialize the discrete random variable.
        
        :param values: A list of possible values of the random variable.
        :param probabilities: A list of probabilities associated with each value.
        """
        if len(values) != len(probabilities):
            raise ValueError("The length of values and probabilities must be the same.")
        
        if not all(0 <= p <= 1 for p in probabilities):
            raise ValueError("Probabilities must be between 0 and 1.")
        
        if sum(probabilities) != 1:
            raise ValueError("The sum of probabilities must equal 1.")
        
        self.values = values
        self.probabilities = probabilities
    
    def expected_value(self):
        """
        Calculate the expected value (mean) of the random variable.
        
        :return: The expected value.
        """
        return sum(value * prob for value, prob in zip(self.values, self.probabilities))
    
    def variance(self):
        """
        Calculate the variance of the random variable.
        
        :return: The variance.
        """
        mean = self.expected_value()
        return sum(prob * (value - mean) ** 2 for value, prob in zip(self.values, self.probabilities))
    
    def __repr__(self):
        return f"DiscreteRandomVariable(values={self.values}, probabilities={self.probabilities})"
# Define the possible values of the random variable and their corresponding probabilities
values = [1, 2, 3, 4, 5]
probabilities = [0.1, 0.2, 0.3, 0.2, 0.2]

# Create a DiscreteRandomVariable instance
drv = DiscreteRandomVariable(values, probabilities)

# Calculate the expected value
expected_value = drv.expected_value()
print(f"Expected Value: {expected_value}")

# Calculate the variance
variance = drv.variance()
print(f"Variance: {variance}")
```

### 4) Implement a program to simulate the rolling of a fair six-sided die and calculate the expected value and variance of the outcomes.

```python
import random
from collections import Counter

class DiscreteRandomVariable:
    def __init__(self, values, probabilities):
        self.values = values
        self.probabilities = probabilities
    
    def expected_value(self):
        return sum(value * prob for value, prob in zip(self.values, self.probabilities))
    
    def variance(self):
        mean = self.expected_value()
        return sum(prob * (value - mean) ** 2 for value, prob in zip(self.values, self.probabilities))


def simulate_dice_rolls(num_rolls):
    outcomes = []
    
    # Simulate the dice rolls
    for _ in range(num_rolls):
        roll1 = random.randint(1, 6)
        roll2 = random.randint(1, 6)
        outcome = roll1 + roll2
        outcomes.append(outcome)
    
    return outcomes

def calculate_probabilities(outcomes):
    total_rolls = len(outcomes)
    outcome_counts = Counter(outcomes)
    
    # Calculate the probability of each outcome
    probabilities = {outcome: count / total_rolls for outcome, count in outcome_counts.items()}
    
    # Ensure the probabilities cover all outcomes from 2 to 12
    all_outcomes = list(range(2, 13))
    probabilities_list = [probabilities.get(outcome, 0) for outcome in all_outcomes]
    
    return all_outcomes, probabilities_list


# Number of simulations
num_rolls = 100000

# Simulate the rolls
outcomes = simulate_dice_rolls(num_rolls)

# Calculate the probabilities
values, probabilities = calculate_probabilities(outcomes)

# Create a DiscreteRandomVariable instance
dice_rv = DiscreteRandomVariable(values, probabilities)

# Calculate the expected value
expected_value = dice_rv.expected_value()
print(f"Expected Value: {expected_value}")

# Calculate the variance
variance = dice_rv.variance()
print(f"Variance: {variance}")
```

### 5) Create a Python function to generate random samples from a given probability distribution (e.g., binomial, Poisson) and calculate their mean and variance.

```python
import numpy as np

def generate_samples_and_calculate_stats(distribution, params, sample_size):
    """
    Generate random samples from a specified distribution and calculate the mean and variance.
    
    :param distribution: The name of the distribution ('binomial', 'poisson', etc.)
    :param params: A dictionary of parameters required for the distribution.
    :param sample_size: The number of random samples to generate.
    :return: A tuple containing the mean and variance of the generated samples.
    """
    if distribution == 'binomial':
        n = params.get('n', 1)  # Number of trials
        p = params.get('p', 0.5)  # Probability of success
        samples = np.random.binomial(n, p, sample_size)
    
    elif distribution == 'poisson':
        lam = params.get('lam', 1)  # Rate (lambda) parameter
        samples = np.random.poisson(lam, sample_size)
    
    else:
        raise ValueError("Unsupported distribution. Supported: 'binomial', 'poisson'")
    
    # Calculate mean and variance of the samples
    mean = np.mean(samples)
    variance = np.var(samples)
    
    return mean, variance

# Example usage:
sample_size = 10000

# Binomial distribution example
binomial_params = {'n': 10, 'p': 0.5}
binomial_mean, binomial_variance = generate_samples_and_calculate_stats('binomial', binomial_params, sample_size)
print(f"Binomial Distribution -> Mean: {binomial_mean}, Variance: {binomial_variance}")

# Poisson distribution example
poisson_params = {'lam': 4}
poisson_mean, poisson_variance = generate_samples_and_calculate_stats('poisson', poisson_params, sample_size)
print(f"Poisson Distribution -> Mean: {poisson_mean}, Variance: {poisson_variance}")
```

```python
import numpy as np
from scipy.stats import norm

def calculate_pdf(x, mu, sigma):
    """
    Calculate the Probability Density Function (PDF) of a normal distribution.

    Parameters:
    x (float): The value at which to evaluate the PDF.
    mu (float): The mean of the normal distribution.
    sigma (float): The standard deviation of the normal distribution.

    Returns:
    float: The PDF value at x.
    """
    pdf_value = norm.pdf(x, loc=mu, scale=sigma)
    return pdf_value

x = 1.5  
mu = 0    
sigma = 1 
pdf_value = calculate_pdf(x, mu, sigma)
print(f"The PDF value at x = {x} for a normal distribution with mean = {mu} and sigma = {sigma} is {pdf_value}")
```

### 15)A company is testing two different website layouts to see which one leads to higher click-through rates.Write a Python function to perform an A/B test analysis, including calculating the t-statistic, degrees of freedom, and p-value.

```python
import numpy as np
from scipy import stats

def ab_test_analysis(layout_a_clicks, layout_b_clicks):
    # Convert the lists to numpy arrays
    a = np.array(layout_a_clicks)
    b = np.array(layout_b_clicks)
    
    # Calculate the means of both layouts
    mean_a = np.mean(a)
    mean_b = np.mean(b)
    
    # Calculate the standard deviations of both layouts
    std_a = np.std(a, ddof=1)
    std_b = np.std(b, ddof=1)
    
    # Calculate the number of samples in each layout
    n_a = len(a)
    n_b = len(b)
    
    # Calculate the standard error of the difference in means
    se = np.sqrt((std_a ** 2) / n_a + (std_b ** 2) / n_b)
    
    # Calculate the t-statistic
    t_statistic = (mean_a - mean_b) / se
    
    # Calculate the degrees of freedom
    df = n_a + n_b - 2
    
    # Calculate the p-value using the t-distribution
    p_value = stats.t.cdf(t_statistic, df=df) * 2  # two-tailed test
    
    return t_statistic, df, p_value

# Given data
layout_a_clicks = [28, 32, 33, 29, 31, 34, 30, 35, 36, 37]
layout_b_clicks = [40, 41, 38, 42, 39, 44, 43, 41, 45, 47]

# Perform the A/B test analysis
t_statistic, df, p_value = ab_test_analysis(layout_a_clicks, layout_b_clicks)

# Output the results
t_statistic, df, p_value
```

### 16) A pharmaceutical company wants to determine if a new drug is more effective than an existing drug in reducing cholesterol levels. Create a program to analyze the clinical trial data and calculate the tstatistic and p-value for the treatment effect.

```python
import numpy as np
from scipy import stats

def analyze_drug_effect(existing_drug_levels, new_drug_levels):
    # Convert the lists to numpy arrays
    existing = np.array(existing_drug_levels)
    new = np.array(new_drug_levels)
        # Calculate the means of both groups
    mean_existing = np.mean(existing)
    mean_new = np.mean(new)   
    # Calculate the standard deviations of both groups
    std_existing = np.std(existing, ddof=1)
    std_new = np.std(new, ddof=1)
    
    # Calculate the number of samples in each group
    n_existing = len(existing)
    n_new = len(new)    
    # Calculate the standard error of the difference in means
    se = np.sqrt((std_existing ** 2) / n_existing + (std_new ** 2) / n_new)
    # Calculate the t-statistic
    t_statistic = (mean_existing - mean_new) / se
    # Calculate the degrees of freedom using Welch's formula
    df_numerator = (std_existing ** 2 / n_existing + std_new ** 2 / n_new) ** 2
    df_denominator = ((std_existing ** 2 / n_existing) ** 2 / (n_existing - 1) +
                      (std_new ** 2 / n_new) ** 2 / (n_new - 1))
    df = df_numerator / df_denominator
 # Calculate the p-value for the t-statistic (two-tailed test)
    p_value = stats.t.sf(np.abs(t_statistic), df=df) * 2
    
    return t_statistic, df, p_value
# Given data
existing_drug_levels = [180, 182, 175, 185, 178, 176, 172, 184, 179, 183]
new_drug_levels = [170, 172, 165, 168, 175, 173, 170, 178, 172, 176]
# Perform the analysis
t_statistic, df, p_value = analyze_drug_effect(existing_drug_levels, new_drug_levels)
# Output the results
t_statistic, df, p_value
```

```python
import numpy as np
from scipy import stats

def analyze_intervention(pre_intervention_scores, post_intervention_scores):
    # Convert the lists to numpy arrays
    pre = np.array(pre_intervention_scores)
    post = np.array(post_intervention_scores)
    
    # Ensure the lengths of both arrays are the same
    if len(pre) != len(post):
        raise ValueError("The lengths of pre- and post-intervention scores must be the same.")
    
    # Calculate the differences between post and pre scores
    differences = post - pre
    
    # Calculate the mean and standard deviation of the differences
    mean_diff = np.mean(differences)
    std_diff = np.std(differences, ddof=1)
    
    # Calculate the number of pairs
    n = len(differences)
    
    # Calculate the standard error of the mean difference
    se_diff = std_diff / np.sqrt(n)
    
    # Calculate the t-statistic
    t_statistic = mean_diff / se_diff
    
    # Calculate the degrees of freedom
    df = n - 1
    
    # Calculate the p-value for the t-statistic (two-tailed test)
    p_value = stats.t.sf(np.abs(t_statistic), df=df) * 2
    
    return t_statistic, df, p_value

# Given data
pre_intervention_scores = [80, 85, 90, 75, 88, 82, 92, 78, 85, 87]
post_intervention_scores = [90, 92, 88, 92, 95, 91, 96, 93, 89, 93]

# Perform the analysis
t_statistic, df, p_value = analyze_intervention(pre_intervention_scores, post_intervention_scores)

# Output the results
t_statistic, df, p_value
```

```python
import numpy as np
from scipy import stats

# Generate synthetic salary data for male and female employees
np.random.seed(0)  # For reproducibility

male_salaries = np.random.normal(loc=50000, scale=10000, size=20)
female_salaries = np.random.normal(loc=55000, scale=9000, size=20)

def analyze_salary_gap(male_salaries, female_salaries):
    # Convert the lists to numpy arrays
    male = np.array(male_salaries)
    female = np.array(female_salaries)
    
    # Calculate the means of both groups
    mean_male = np.mean(male)
    mean_female = np.mean(female)
    
    # Calculate the standard deviations of both groups
    std_male = np.std(male, ddof=1)
    std_female = np.std(female, ddof=1)
    
    # Calculate the number of samples in each group
    n_male = len(male)
    n_female = len(female)
    
    # Calculate the standard error of the difference in means
    se = np.sqrt((std_male ** 2) / n_male + (std_female ** 2) / n_female)
    
    # Calculate the t-statistic
    t_statistic = (mean_male - mean_female) / se
    
    # Calculate the degrees of freedom using Welch's formula
    df_numerator = (std_male ** 2 / n_male + std_female ** 2 / n_female) ** 2
    df_denominator = ((std_male ** 2 / n_male) ** 2 / (n_male - 1) +
                      (std_female ** 2 / n_female) ** 2 / (n_female - 1))
    df = df_numerator / df_denominator
    
    # Calculate the p-value for the t-statistic (two-tailed test)
    p_value = stats.t.sf(np.abs(t_statistic), df=df) * 2
    
    return t_statistic, df, p_value

# Perform the analysis
t_statistic, df, p_value = analyze_salary_gap(male_salaries, female_salaries)

# Output the results
print(f"T-statistic: {t_statistic:.3f}")
print(f"Degrees of freedom: {df:.3f}")
print(f"P-value: {p_value:.3f}")
```

### 21)A political analyst wants to determine if there is a significant association between age groups and voter preferences FCandidate A or Candidate B). They collect data from a sample of 500 voters and classify them into different age groups and candidate preferences. Perform a Chi-Square test to determine if there is a significant association between age groups and voter preferences.

```python
import numpy as np
import pandas as pd
from scipy import stats

# Generate synthetic data
np.random.seed(0)

age_groups = np.random.choice(['18-30', '31-50', '51+'], size=500)
voter_preferences = np.random.choice(['Candidate A', 'Candidate B'], size=500)

# Create a DataFrame to organize the data
data = pd.DataFrame({'Age Group': age_groups, 'Voter Preference': voter_preferences})

# Create a contingency table
contingency_table = pd.crosstab(data['Age Group'], data['Voter Preference'])

# Perform the Chi-Square test
chi2_statistic, p_value, df, expected = stats.chi2_contingency(contingency_table)

# Output the results
print("Contingency Table:")
print(contingency_table)
print("\nChi-Square Statistic:", chi2_statistic)
print("Degrees of Freedom:", df)
print("P-Value:", p_value)
```
